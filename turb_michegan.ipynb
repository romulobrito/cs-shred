```python
import matplotlib.pyplot as plt
import seaborn as sns

from mpl_toolkits.mplot3d import Axes3D

nx = 257
ny = 256

fig = plt.figure(figsize=(15, 15))
for k in range(6):
    Up = u_total[:, k].reshape((ny, nx), order='F')
    uplot = Up.T
    ax = fig.add_subplot(3, 3, k+1)
    ax.axis('off')
    ax.imshow(uplot,cmap='RdBu_r', interpolation='bilinear',)
    plt.savefig('u.pdf')
```

```python
# Reorganizar e salvar os dados
import numpy as np

# Reorganizar os dados para o formato esperado pelo piv.py
u_snap = u_total.reshape((256, 257, 280), order='F')  # Mantendo suas dimensões originais

# Criar o diretório se não existir
import os
save_dir = '/home/romulo/Documentos/lpips-env/data/'
os.makedirs(save_dir, exist_ok=True)

# Salvar o array
np.save(os.path.join(save_dir, 'u_total.npy'), u_snap)

# Verificar se salvou corretamente
# Plotar um snapshot para confirmar
plt.figure(figsize=(10, 8))
plt.imshow(u_snap[:,:,1].T, cmap='RdBu_r', interpolation='bilinear')
plt.colorbar()
plt.title('Snapshot 1')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()

# Imprimir as dimensões para confirmar
print("Dimensões do array salvo:", u_snap.shape)
```
